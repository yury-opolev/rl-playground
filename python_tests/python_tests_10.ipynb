```python
import numpy as np
import tensorflow as tf

from tensorflow import keras
from keras import layers

print(f"tensorflow version: {tf.__version__}, keras version: {keras.__version__}")
```

```python
inputs = keras.Input(shape=(2,))
x_1 = layers.Dense(2, activation="leaky_relu")(inputs)
x_1 = layers.Dense(1, activation="leaky_relu")(x_1)

x_2 = layers.Dense(2, activation="leaky_relu")(inputs)
x_2 = layers.Dense(1, activation="leaky_relu")(x_2)

outputs = y = keras.layers.Concatenate()([x_1, x_2])

model = keras.Model(inputs=inputs, outputs=outputs)
```

```python
keras.utils.plot_model(model, "my_first_model.png", show_shapes=True)
```
