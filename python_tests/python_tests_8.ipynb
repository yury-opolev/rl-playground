```python
input_1 = tf.Variable([[1.0, 1.0]])
expected_y_1 = tf.Variable([[0.17137074, -0.5]])

print(f"input: {input_1}")
print(f"expected_output: {expected_y_1}")

with tf.GradientTape(persistent=True) as tape:
    output_1 = model(input_1)
    loss = tf.abs(expected_y_1 - output_1)

gradients = tape.gradient(loss, model.trainable_weights)

print(f"output: {output_1}")
print(f"loss: {loss}")
print(f"gradients: {gradients}")

for layer in model.layers:
    print(f"layer {layer.name} weights: {layer.get_weights()}")
```

```python
optimizer = keras.optimizers.SGD(learning_rate=0.1)
optimizer.apply_gradients(zip(gradients, model.trainable_weights))

for layer in model.layers:
    print(f"layer {layer.name} weights: {layer.get_weights()}")
```
