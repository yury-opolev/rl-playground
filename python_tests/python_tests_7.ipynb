```python
import random
import tensorflow as tf
import keras
import numpy as np

from keras import layers
from keras import models
from keras import initializers

print(f"tensorflow version: {tf.__version__}, keras version: {keras.__version__}")
```

```python
nn_model = models.Sequential([
            layers.Input(shape=(2,)),
            layers.Dense(2, activation=keras.activations.leaky_relu,
                         kernel_initializer=initializers.RandomNormal(stddev=0.05),
                         bias_initializer=initializers.RandomNormal(stddev=0.05)),
            layers.Dense(2, activation=keras.activations.linear,
                         kernel_initializer=initializers.RandomNormal(stddev=0.05),
                         bias_initializer=initializers.RandomNormal(stddev=0.05))
        ])
```

```python
input_state_1 = tf.convert_to_tensor([[1.0, 0.0]])

output_1 = nn_model(input_state_1)

print(f"output: {output_1}")

expected_output_1 = tf.Variable(initial_value=output_1, trainable=False)

temp = expected_output_1.numpy()
temp[0][1] = 0.75

expected_output_1.assign(temp)

print(f"expected output: {expected_output_1.numpy()}")

```

```python
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)

with tf.GradientTape() as tape:
    predicted_values = nn_model(input_state_1, training=True)
    diff = tf.slice(tf.subtract(expected_output_1, predicted_values), [0,1], [-1,1])
    loss = tf.abs(diff)
    print(f"loss: {loss}")

gradients = tape.gradient(loss, nn_model.trainable_weights)

print(f"gradients: {gradients}")

optimizer.apply_gradients(zip(gradients, nn_model.trainable_variables))

#for i, gradient in enumerate(gradients):
#    nn_model.trainable_weights[i].assign_sub(learning_rate * gradient)
```
